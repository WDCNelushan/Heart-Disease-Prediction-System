```python
# Detect ouliers

import seaborn as sns

plt.figure(figsize=(12, 6))
sns.boxplot(data=data[['Age', 'RestingBp', 'Cholesterol', 'MaxHR']])
plt.title('Boxplot for Outlier Detection')
plt.show()
```

```python
# Handling Outliers
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Applying the function to remove outliers from specific columns
columns_to_check = ['Age', 'RestingBp', 'Cholesterol', 'MaxHR']
for column in columns_to_check:
    data = remove_outliers(data, column)


```

```python
# Data preprocessing

# Seperating features and target variable
# encoding categorical variables
# Scalling numerical features
# Splitting the dataset into training and testing sets
```

```python
# Seperating features and target variable
X = data.drop('HeartDisease', axis=1)
y = data['HeartDisease']
print("Features shape:", X.shape)
```

```python
# Encoding categorical variables
categorical_columns = ['Gender', 'ChestPainType', 'FastingBS', 'RestingECG', 'ExerciseEngage', 'ST_Slope', 'MajorVessels', 'Thalassemia']
numerical_columns = ['Age', 'RestingBp', 'Cholesterol', 'MaxHR', 'ST_Depression']
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split

#encoding categorical variables using OneHotEncoder
x_encoded = pa.get_dummies(data, columns=categorical_columns, drop_first=True)
print("Encoded features shape:", x_encoded.shape)
x_encoded.head()
```

```python
# feature Scaling using StandardScaler
scaler = StandardScaler()
x_encoded[numerical_columns] = scaler.fit_transform(x_encoded[numerical_columns])
print("Scaled numerical features")
print(x_encoded[numerical_columns].head())
```

```python
#Splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(x_encoded,y, test_size=0.2, random_state=42,stratify=y)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)
```

```python
# Model Training and Evaluation

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier 
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Logistic Regression Model
log_rg = LogisticRegression()
log_rg.fit(X_train, y_train)
y_pred_log = log_rg.predict(X_test)
accuracy_log = accuracy_score(y_test, y_pred_log)
print("Logistic Regression Accuracy:", accuracy_log)
cm_log_rg = confusion_matrix(y_test, y_pred_log)
print("Confusion Matrix for Logistic Regression:\n", cm_log_rg)
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(cm_log_rg, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix for Logistic Regression')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Random Forest Classifier Model
rf_clf = RandomForestClassifier(n_estimators=200, random_state=42)
rf_clf.fit(X_train, y_train)
y_pred_rf = rf_clf.predict(X_test)

accuracy_rf = accuracy_score(y_test, y_pred_rf)
cm_rf = confusion_matrix(y_test, y_pred_rf)
print("Classification Report for Random Forest Classifier:\n", classification_report(y_test, y_pred_rf))
print("Confusion Matrix for Random Forest Classifier:\n", cm_rf)
print("Accuracy for Random Forest Classifier:", round(accuracy_rf *100, 2),'%')




```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Greens')
plt.title('Confusion Matrix for Random Forest')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Comparing the performance of both models
results = pa.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest Classifier'],
    'Accuracy': [accuracy_log, accuracy_rf]
})

print("Model Comparison:\n", results)

plt.figure(figsize=(8, 5))
sns.barplot(x='Model', y='Accuracy', data=results,palette='viridis')
plt.title('Model Comparison')
plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.show()
```
